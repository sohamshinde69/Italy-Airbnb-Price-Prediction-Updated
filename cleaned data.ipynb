```python
import pandas as pd 
import numpy as np
```

```python
rome_url = "https://data.insideairbnb.com/italy/lazio/rome/2026-06-20/data/listings.csv.gz"
```

```python
rome_all = pd.read_csv(rome_url, compression="gzip", nrows=5)

print(rome_all.columns.tolist())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
```

```python
numeric_cols = [
    "accommodates",
    "bathrooms",
    "bedrooms",
    "minimum_nights",
    "number_of_reviews",
    "review_scores_rating"
]
```

```python
categorical_cols = [
    "property_type",
    "room_type",
    "city"
]
```

```python
X = final_data[numeric_cols + categorical_cols]
y = final_data["price"]
```

```python
ct = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ), categorical_cols),
        
    ],
    remainder = "passthrough"
)
```

```python
X_transformed = ct.fit_transform(X)
```

```python
print(X.shape)
print(X_transformed.shape)
```

```python
feature_names = ct.get_feature_names_out()
print(feature_names)
print(len(feature_names))
```

```python
from sklearn.model_selection import train_test_split


X_train , X_test , y_train , y_test = train_test_split(
    X, y, test_size=0.23, random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline = Pipeline([
    ("preprocessor", ct),
    ("model", RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
rf_pipeline.fit(X_train , y_train)
```

```python
y_pred_rf = rf_pipeline.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)
```

```python
print("Random Forest - MAE:", mae_rf)
print("Random Forest - RMSE:", rmse_rf)
print("Random Forest - R²:", r2_rf)
```

```python
from sklearn.linear_model import LinearRegression

lr_pipeline = Pipeline([
    ("preprocessor", ct),
    ("model", LinearRegression())
])
```

```python
lr_pipeline.fit(X_train, y_train)
```

```python
y_pred_lr = lr_pipeline.predict(X_test)
```

```python
mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
r2_lr = r2_score(y_test, y_pred_lr)
```

```python
print("Linear Regression - MAE:", mae_lr)
print("Linear Regression - RMSE:", rmse_lr)
print("Linear Regression - R²:", r2_lr)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gbr_pipeline = Pipeline([
    ("preprocessor", ct),
    ("model", GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ))
])
```

```python
gbr_pipeline.fit(X_train, y_train)
```

```python
y_pred_gbr = gbr_pipeline.predict(X_test)
```

```python
mae_gbr = mean_absolute_error(y_test, y_pred_gbr)
rmse_gbr = np.sqrt(mean_squared_error(y_test, y_pred_gbr))
r2_gbr = r2_score(y_test, y_pred_gbr)
```

```python
print("Gradient Boosting - MAE:", mae_gbr)
print("Gradient Boosting - RMSE:", rmse_gbr)
print("Gradient Boosting - R²:", r2_gbr)
```
